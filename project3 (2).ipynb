# Exploratry Data Analysis (EDA)

## Declare Chategorical data

```python
# Define category and continuous columns are sepratly 
```

```python
categorical = []  

for i in df:
    if len(df[i].unique())<=20:
        categorical.append(i)
        
    
```

```python
df_cat = df[categorical]
df_cat.head()
```

```python
categorical
```

## Declare Numerical/Continuos Data

```python
continuous = []

for i in df:
    if len(df[i].unique())>20:
        continuous.append(i)
```

```python
df_cont = df[continuous]
df_cont.head()
```

```python
continuous 
```

### Following are main two Analysis of (EDA)

```python
# Fiestly Define univariate Analysis :- univariate analysis only one columns are store categorical/countinuos in this called
# Analysis
```

## Univeriate Analysis (Categorical)

```python
plotnumber = 1
plt.figure(figsize=(16,22))

for i in df_cat :
    plt.subplot(6,5,plotnumber)
    sns.countplot(x=df_cat[i])
    plt.xticks(rotation = 45)
    plotnumber = plotnumber +1
    plt.tight_layout()
```

## univeriate Analysis (Continuous)

```python
plotnumber = 1
plt.figure(figsize=(10,16))

for i in df_cont :
    plt.subplot(5,4,plotnumber)
    sns.histplot(x=df_cont[i],kde=True)
    plt.xticks(rotation = 45)
    plotnumber = plotnumber +1
    plt.tight_layout()
```
